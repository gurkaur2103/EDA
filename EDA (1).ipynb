```python
#Column Helper
print("=== COURSERA COLUMNS ===", coursera.columns.tolist())
print("\n=== UDEMY COLUMNS ===",   udemy.columns.tolist())
print("\n=== SKILLSHARE COLUMNS ===", skillshare.columns.tolist())
```

```python
# ── Standardise key column names ───────────────────────────────────────────
# Coursera — map to common names
COURSERA_MAP = {
    'Course Name':  'title',
    'Category':     'category',
    'Level':        'level',
    'Institution':  'institution',
    'Duration':     'duration',
    'Rating':       'rating',
    'Reviews':      'num_reviews',
    'Students':     'enrolled',
}
coursera.rename(columns={k:v for k,v in COURSERA_MAP.items() if k in coursera.columns}, inplace=True)

# Udemy
UDEMY_MAP = {
    'course_title':       'title',
    'subject':            'subject',
    'is_paid':            'is_paid',
    'price':              'price',
    'num_subscribers':    'num_subscribers',
    'num_reviews':        'num_reviews',
    'num_lectures':       'num_lectures',
    'level':              'level',
    'content_duration':   'hours',
    'published_timestamp':'published_at',
    'rating':             'rating',
}
udemy.rename(columns={k:v for k,v in UDEMY_MAP.items() if k in udemy.columns}, inplace=True)

# Skillshare
SKILLSHARE_MAP = {
    'Title':          'title',
    'Category':       'category',
    'Students':       'num_students',
    'Projects':       'num_projects',
    'Lessons':        'num_lessons',
    'Duration':       'duration_min',
    'Rating':         'rating',
    'Premium':        'is_premium',
}
skillshare.rename(columns={k:v for k,v in SKILLSHARE_MAP.items() if k in skillshare.columns}, inplace=True)

# Numeric coercions
for df, cols in [
    (coursera,  ['rating','num_reviews','enrolled']),
    (udemy,     ['rating','price','num_subscribers','num_reviews','num_lectures','hours']),
    (skillshare,['rating','num_students','num_projects','num_lessons']),
]:
    for c in cols:
        if c in df.columns:
            df[c] = pd.to_numeric(df[c], errors='coerce')

print("Cleaning done ✓")
print(f"Rows — Coursera: {len(coursera):,}  |  Udemy: {len(udemy):,}  |  Skillshare: {len(skillshare):,}")
```

```python
# ── Standardize Columns ───────────────────────────────────────────────────

# ---------- Udemy Dataset ----------

udemy.columns = (
    udemy.columns
    .str.strip()
    .str.lower()
    .str.replace(' ', '_')
)

udemy.rename(columns={
    'category': 'category',
    'rating': 'rating',
    'duration': 'duration',
    'number_of_reviews': 'num_reviews',
    'number_of_viewers': 'num_students',
    'price': 'price',
    'level': 'level'
}, inplace=True)

# Convert numeric columns safely
for col in ['rating', 'num_reviews', 'price']:

    if col in udemy.columns:

        udemy[col] = pd.to_numeric(
            udemy[col],
            errors='coerce'
        )

# ---------- Skillshare Dataset ----------

skillshare.columns = (
    skillshare.columns
    .str.strip()
    .str.lower()
    .str.replace(' ', '_')
)

# Extract numeric students
if 'students' in skillshare.columns:

    skillshare['students_count'] = (
        skillshare['students']
        .str.replace(' students', '', regex=False)
        .str.replace(',', '', regex=False)
    )

    skillshare['students_count'] = pd.to_numeric(
        skillshare['students_count'],
        errors='coerce'
    )

# Convert duration into minutes
if 'duration' in skillshare.columns:

    def convert_duration(x):

        if pd.isna(x):
            return np.nan

        x = str(x).lower()

        hours = 0
        mins = 0

        if 'h' in x:
            try:
                hours = int(x.split('h')[0].strip())
            except:
                pass

        if 'm' in x:
            try:
                mins = int(
                    x.split('h')[-1]
                    .replace('m', '')
                    .strip()
                )
            except:
                pass

        return hours * 60 + mins

    skillshare['duration_min'] = skillshare['duration'].apply(convert_duration)

print("Udemy Columns:")
print(udemy.columns.tolist())

print("\nSkillshare Columns:")
print(skillshare.columns.tolist())
```

```python
# ── Missing values heatmap ─────────────────────────────────────────────────
fig, axes = plt.subplots(1, 3, figsize=(18, 4))
for ax, (df, name) in zip(axes, [(coursera,'Coursera'),(udemy,'Udemy'),(skillshare,'Skillshare')]):
    miss = df.isnull().mean().sort_values(ascending=False)
    miss = miss[miss > 0] if miss.any() else miss.head(8)
    colors = [BLUE if v < 0.05 else ORANGE if v < 0.20 else '#e85c6e' for v in miss.values]
    ax.barh(miss.index, miss.values * 100, color=colors)
    ax.set_xlabel('Missing %')
    ax.set_title(f'{name} — Missing Values', fontweight='bold')
    ax.axvline(5,  color='#f5c842', lw=1, ls='--', alpha=0.7, label='5%')
    ax.axvline(20, color='#e85c6e', lw=1, ls='--', alpha=0.7, label='20%')
    ax.legend(fontsize=8)
plt.tight_layout(); plt.show()
```

```python
# ── 3.2  Rating distribution — KDE overlay (all 3 platforms) ──────────────
fig, ax = plt.subplots(figsize=(12, 5))

for df, name, color in [
    (coursera,  'Coursera',   BLUE),
    (udemy,     'Udemy',      ORANGE),
    (skillshare,'Skillshare', GREEN),
]:
    if 'rating' in df.columns:
        data = df['rating'].dropna()
        ax.hist(data, bins=30, density=True, alpha=0.25, color=color)
        kde_x = np.linspace(data.min(), data.max(), 300)
        kde   = stats.gaussian_kde(data)(kde_x)
        ax.plot(kde_x, kde, color=color, lw=2.5, label=f'{name} (μ={data.mean():.2f})')
        ax.axvline(data.mean(), color=color, lw=1.2, ls='--', alpha=0.7)

ax.set_xlabel('Rating'); ax.set_ylabel('Density')
ax.set_title('Rating Distribution — All Platforms', fontweight='bold')
ax.legend(); ax.grid(True, alpha=0.3)
plt.tight_layout(); plt.show()
```

```python
# ── 4.2 Level & Duration ─────────────────────────────────────────────────

fig, axes = plt.subplots(
    1,
    2,
    figsize=(14, 5)
)

plots = [
    (axes[0], 'level', 'Level'),
    (axes[1], 'duration', 'Duration')
]

for ax, col, label in plots:

    # Check column exists
    if col in udemy.columns:

        # Clean values
        vc = (
            udemy[col]
            .dropna()
            .astype(str)
            .str.strip()
            .value_counts()
            .head(10)
        )

        # If no values
        if vc.empty:

            ax.text(
                0.5,
                0.5,
                'No Data',
                ha='center',
                va='center',
                fontsize=12
            )

            ax.axis('off')
            continue

        # Colors
        pie_colors = plt.cm.Set3(
            np.linspace(0, 1, len(vc))
        )

        # Pie chart
        ax.pie(
            vc.values,
            labels=vc.index,
            autopct='%1.1f%%',
            colors=pie_colors,
            startangle=90,
            wedgeprops=dict(
                edgecolor='white',
                linewidth=1.5
            )
        )

        ax.set_title(
            f'Udemy — {label} Distribution',
            fontweight='bold'
        )

    else:

        ax.text(
            0.5,
            0.5,
            f'{col} column missing',
            ha='center',
            va='center',
            fontsize=12
        )

        ax.axis('off')

plt.tight_layout()
plt.show()
```

```python
# ── 4.3 Average enrollment by category ───────────────────────────────────

required_cols = {'category', 'num_students'}

if required_cols.issubset(udemy.columns):

    # Clean numeric values
    udemy['num_students'] = pd.to_numeric(
        udemy['num_students'],
        errors='coerce'
    )

    # Remove missing rows
    temp_df = udemy.dropna(
        subset=['category', 'num_students']
    )

    # Group by category
    enroll_cat = (
        temp_df
        .groupby('category')['num_students']
        .mean()
        .sort_values(ascending=True)
        .tail(10)
    )

    # Create figure
    fig, ax = plt.subplots(figsize=(12, 6))

    colors = plt.cm.viridis(
        np.linspace(0.3, 0.9, len(enroll_cat))
    )

    bars = ax.barh(
        enroll_cat.index,
        enroll_cat.values,
        color=colors
    )

    # Labels
    for bar, val in zip(bars, enroll_cat.values):

        ax.text(
            bar.get_width() + 200,
            bar.get_y() + bar.get_height()/2,
            f'{val:,.0f}',
            va='center',
            fontsize=9
        )

    ax.set_xlabel('Average Enrollment')

    ax.set_title(
        'Udemy — Average Enrollment per Category',
        fontweight='bold'
    )

    ax.grid(axis='x', alpha=0.3)

    plt.tight_layout()
    plt.show()

else:

    print("Required columns missing")
```

```python
# ── 4.4  Rating vs Reviews scatter (Coursera) ──────────────────────────────
if {'rating','num_reviews'}.issubset(coursera.columns):
    sample = coursera.dropna(subset=['rating','num_reviews']).sample(min(500, len(coursera)), random_state=42)

    fig, ax = plt.subplots(figsize=(10, 6))
    sc = ax.scatter(sample['rating'], sample['num_reviews'],
                    c=sample['rating'], cmap='plasma',
                    alpha=0.55, s=40, edgecolors='none')
    plt.colorbar(sc, ax=ax, label='Rating')

    # Regression line
    m, b, r, p, _ = stats.linregress(sample['rating'], sample['num_reviews'])
    x_line = np.linspace(sample['rating'].min(), sample['rating'].max(), 100)
    ax.plot(x_line, m*x_line+b, color=ORANGE, lw=2, label=f'OLS  r={r:.2f}, p={p:.3f}')

    ax.set_xlabel('Rating'); ax.set_ylabel('Number of Reviews')
    ax.set_title('Coursera — Rating vs. Number of Reviews', fontweight='bold')
    ax.legend(); ax.grid(True, alpha=0.3)
    plt.tight_layout(); plt.show()
```

```python
# ── 5.1 Students by category ─────────────────────────────────────────────

required_cols = {'category', 'num_students'}

if required_cols.issubset(udemy.columns):

    # Convert to numeric
    udemy['num_students'] = pd.to_numeric(
        udemy['num_students'],
        errors='coerce'
    )

    # Remove missing rows
    temp_df = udemy.dropna(
        subset=['category', 'num_students']
    )

    # Group data
    sub_enroll = (
        temp_df
        .groupby('category')['num_students']
        .mean()
        .sort_values(ascending=False)
        .head(10)
    )

    # Create figure
    fig, ax = plt.subplots(
        figsize=(12, 6)
    )

    colors = plt.cm.Oranges(
        np.linspace(0.4, 0.9, len(sub_enroll))[::-1]
    )

    bars = ax.bar(
        sub_enroll.index,
        sub_enroll.values,
        color=colors,
        zorder=2
    )

    # Labels
    for bar, val in zip(bars, sub_enroll.values):

        ax.text(
            bar.get_x() + bar.get_width()/2,
            bar.get_height() + 100,
            f'{val:,.0f}',
            ha='center',
            va='bottom',
            fontsize=8,
            rotation=45
        )

    ax.set_xticks(range(len(sub_enroll.index)))
    ax.set_xticklabels(
        sub_enroll.index,
        rotation=30,
        ha='right'
    )

    ax.set_ylabel('Average Students')

    ax.set_title(
        'Udemy — Average Students per Category',
        fontweight='bold'
    )

    ax.grid(
        axis='y',
        alpha=0.3,
        zorder=1
    )

    plt.tight_layout()
    plt.show()

else:

    print("Required columns missing")
```

```python
# ── Better duration parser ───────────────────────────────────────────────

import re

def convert_duration(x):

    if pd.isna(x):
        return np.nan

    x = str(x).lower()

    # Months
    month_match = re.search(r'(\d+)\s*month', x)

    if month_match:
        months = int(month_match.group(1))

        # Approximation:
        # 1 month ≈ 30 days × 2 study hours/day
        return months * 60

    # Weeks
    week_match = re.search(r'(\d+)\s*week', x)

    if week_match:
        weeks = int(week_match.group(1))
        return weeks * 14

    # Hours
    hour_match = re.search(r'(\d+)\s*h', x)

    mins_match = re.search(r'(\d+)\s*m', x)

    hours = int(hour_match.group(1)) if hour_match else 0
    mins  = int(mins_match.group(1)) if mins_match else 0

    return hours + mins / 60

# Apply conversion
udemy['hours'] = udemy['duration'].apply(convert_duration)

# Check results
print(
    udemy[['duration', 'hours']].head(10)
)
```

```python
# ── 7.3 Platform popularity distribution ────────────────────────────────
import re
fig, ax = plt.subplots(figsize=(12, 7))

# Helper function
def get_students_column(df):

    possible_cols = [
        'num_students',
        'enrolled',
        'students',
        'num_enrolled',
        'learners',
        'number_of_viewers'
    ]

    for col in possible_cols:

        if col in df.columns:
            return col

    return None
# ───────────────────────────────────────────
# Coursera
# ───────────────────────────────────────────

def parse_enrollment(value):

    if pd.isna(value):
        return np.nan

    value = str(value).lower().replace(',', '').strip()

    try:

        if 'm' in value:
            return float(value.replace('m', '')) * 1_000_000

        elif 'k' in value:
            return float(value.replace('k', '')) * 1_000

        else:
            return float(
                re.findall(r'[\d.]+', value)[0]
            )

    except:
        return np.nan

# Clean enrolled column
c_students = coursera['enrolled'].apply(parse_enrollment)

c_students = c_students.dropna()

c_students = c_students[
    c_students > 0
]

print("Coursera rows:", len(c_students))

if len(c_students) > 0:

    y = np.log1p(
        c_students.sample(
            min(300, len(c_students)),
            random_state=42
        )
    )

    x = np.random.normal(
        1,
        0.05,
        size=len(y)
    )

    ax.scatter(
        x,
        y,
        alpha=0.35,
        s=40,
        color=BLUE,
        label='Coursera'
    )
# ───────────────────────────────────────────
# Udemy
# ───────────────────────────────────────────

u_col = get_students_column(udemy)

if u_col:

    u_students = pd.to_numeric(
        udemy[u_col],
        errors='coerce'
    ).dropna()

    u_students = u_students[
        u_students > 0
    ]

    print("Udemy column:", u_col)
    print("Udemy rows:", len(u_students))

    if len(u_students) > 0:

        y = np.log1p(
            u_students.sample(
                min(300, len(u_students)),
                random_state=42
            )
        )

        x = np.random.normal(
            2,
            0.05,
            size=len(y)
        )

        ax.scatter(
            x,
            y,
            alpha=0.35,
            s=40,
            color=ORANGE,
            label='Udemy'
        )

# ───────────────────────────────────────────
# Skillshare
# ───────────────────────────────────────────

if 'students_count' in skillshare.columns:

    s_students = pd.to_numeric(
        skillshare['students_count'],
        errors='coerce'
    ).dropna()

    s_students = s_students[
        s_students > 0
    ]

    print("Skillshare rows:", len(s_students))

    if len(s_students) > 0:

        y = np.log1p(
            s_students.sample(
                min(300, len(s_students)),
                random_state=42
            )
        )

        x = np.random.normal(
            3,
            0.05,
            size=len(y)
        )

        ax.scatter(
            x,
            y,
            alpha=0.35,
            s=40,
            color=GREEN,
            label='Skillshare'
        )

# ───────────────────────────────────────────
# Styling
# ───────────────────────────────────────────

ax.set_xticks([1, 2, 3])

ax.set_xticklabels([
    'Coursera',
    'Udemy',
    'Skillshare'
])

ax.set_ylabel(
    'log(1 + Students / Enrollment)'
)

ax.set_title(
    'Platform Popularity Distribution',
    fontweight='bold'
)

ax.grid(
    axis='y',
    alpha=0.3
)

ax.legend()

plt.tight_layout()
plt.show()
```

```python
# ── 7.4 Correlation heatmap per platform ────────────────────────────────

fig, axes = plt.subplots(
    1,
    3,
    figsize=(20, 6)
)

datasets = [
    (coursera, 'Coursera'),
    (udemy, 'Udemy'),
    (skillshare, 'Skillshare')
]

for ax, (df, name) in zip(axes, datasets):

    temp_df = df.copy()

    # Try converting object columns to numeric
    for col in temp_df.columns:

        if temp_df[col].dtype == 'object':

            cleaned = (
                temp_df[col]
                .astype(str)
                .str.replace(',', '', regex=False)
                .str.extract(r'([\d.]+)')[0]
            )

            converted = pd.to_numeric(
                cleaned,
                errors='coerce'
            )

            # Keep conversion only if enough numeric values exist
            if converted.notna().sum() > len(df) * 0.3:

                temp_df[col] = converted

    # Select numeric columns
    num_df = temp_df.select_dtypes(
        include='number'
    )

    # Remove fully empty columns
    num_df = num_df.dropna(
        axis=1,
        how='all'
    )

    # Remove constant columns
    num_df = num_df.loc[
        :,
        num_df.nunique() > 1
    ]

    # Check if enough columns exist
    if num_df.shape[1] < 2:

        ax.text(
            0.5,
            0.5,
            'Not enough numeric data',
            ha='center',
            va='center',
            fontsize=14
        )

        ax.set_title(
            f'{name} — Correlation Matrix',
            fontweight='bold'
        )

        ax.axis('off')

        continue

    # Correlation
    corr = num_df.corr()

    # Mask upper triangle
    mask = np.triu(
        np.ones_like(
            corr,
            dtype=bool
        )
    )

    # Heatmap
    sns.heatmap(
        corr,
        ax=ax,
        mask=mask,
        annot=True,
        fmt='.2f',
        cmap='coolwarm',
        center=0,
        linewidths=0.4,
        linecolor='#0f1117',
        vmin=-1,
        vmax=1,
        cbar_kws={'shrink': 0.8}
    )

    ax.set_title(
        f'{name} — Correlation Matrix',
        fontweight='bold'
    )

plt.tight_layout()
plt.show()
```

```python
# ── 8.1 Plotly Treemap — Coursera organizations by reviews ─────────────

temp_df = coursera.copy()

# Clean num_reviews
temp_df['reviews_clean'] = (
    temp_df['num_reviews']
    .astype(str)
    .str.replace(',', '', regex=False)
    .str.extract(r'(\d+\.?\d*)')[0]
)

temp_df['reviews_clean'] = pd.to_numeric(
    temp_df['reviews_clean'],
    errors='coerce'
)

# Remove invalid rows
temp_df = temp_df.dropna(
    subset=[
        'Organization',
        'reviews_clean'
    ]
)

# Keep positive values
temp_df = temp_df[
    temp_df['reviews_clean'] > 0
]

print("Valid rows:", len(temp_df))

# Top organizations
top_orgs = (
    temp_df['Organization']
    .value_counts()
    .head(15)
    .index
)

temp_df = temp_df[
    temp_df['Organization'].isin(top_orgs)
]

# Grouping
tm = (
    temp_df
    .groupby('Organization')
    .agg(
        courses=('title', 'count'),
        avg_reviews=('reviews_clean', 'mean')
    )
    .reset_index()
)

# Treemap
fig = px.treemap(
    tm,
    path=['Organization'],
    values='courses',
    color='avg_reviews',
    color_continuous_scale='Blues',
    title='Coursera — Organization Treemap (by Reviews)',
    template='plotly_white'
)

fig.update_layout(
    margin=dict(
        t=50,
        l=10,
        r=10,
        b=10
    )
)

fig.show()
```

```python
# ── 8.2 Plotly Scatter — Udemy price vs viewers ────────────────────────

temp_df = udemy.copy()

# Helper to find columns ignoring case
def find_col(df, target):

    for col in df.columns:

        if col.lower() == target.lower():

            return col

    return None

# Detect columns
price_col = find_col(temp_df, 'price')
viewer_col = find_col(temp_df, 'number of viewers')
rating_col = find_col(temp_df, 'rating')
category_col = find_col(temp_df, 'category')
title_col = find_col(temp_df, 'title')

print(price_col)
print(viewer_col)
print(rating_col)
print(category_col)
print(title_col)

# Continue only if columns exist
if all([
    price_col,
    viewer_col,
    rating_col,
    category_col,
    title_col
]):

    # Clean price
    temp_df['price_clean'] = pd.to_numeric(
        temp_df[price_col],
        errors='coerce'
    )

    # Clean viewers
    temp_df['viewers_clean'] = (
        temp_df[viewer_col]
        .astype(str)
        .str.replace(',', '', regex=False)
        .str.extract(r'(\d+\.?\d*)')[0]
    )

    temp_df['viewers_clean'] = pd.to_numeric(
        temp_df['viewers_clean'],
        errors='coerce'
    )

    # Clean rating
    temp_df['rating_clean'] = pd.to_numeric(
        temp_df[rating_col],
        errors='coerce'
    )

    # Remove invalid rows
    temp_df = temp_df.dropna(
        subset=[
            'price_clean',
            'viewers_clean',
            'rating_clean'
        ]
    )

    temp_df = temp_df[
        temp_df['viewers_clean'] > 0
    ]

    print("Valid rows:", len(temp_df))

    # Sample
    sample_p = temp_df.sample(
        min(800, len(temp_df)),
        random_state=7
    )

    # Plot
    fig = px.scatter(
        sample_p,
        x='price_clean',
        y='viewers_clean',
        size='rating_clean',
        color=category_col,
        hover_name=title_col,
        size_max=30,
        opacity=0.65,
        log_y=True,
        title='Udemy — Price vs Viewers',
        template='plotly_white'
    )

    fig.update_layout(
        xaxis_title='Price',
        yaxis_title='Number of Viewers'
    )

    fig.show()

else:

    print("Required columns missing")
```
